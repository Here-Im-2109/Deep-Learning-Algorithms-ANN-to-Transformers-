```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('DailyDelhiClimate.csv')
```

```python
df = df[['date', 'meantemp']]
```

```python
df
```

```python
plt.scatter(df['date'] ,df['meantemp'] ,color = 'red' ,s=10)
```

```python
x1 = df['meantemp'].ewm(alpha = 0.9).mean()
x1
df['ewm']=x1
df
```

```python
plt.scatter(df['date'] ,df['meantemp'] ,color = 'red' ,s=10)
plt.plot(df['date'] ,x1,color = 'black')
```

```python
x2 = df['meantemp'].ewm(alpha = 0.7).mean()
x2
df['ewm']=x2
df
```

```python
plt.scatter(df['date'] ,df['meantemp'] ,color = 'red' ,s=10)
plt.plot(df['date'] ,x2,color = 'black')
```

```python
x3 = df['meantemp'].ewm(alpha = 0.5).mean()
x3
df['ewm']=x3
df
```

```python
plt.scatter(df['date'] ,df['meantemp'] ,color = 'red' ,s=10)
plt.plot(df['date'] ,x3,color = 'black')
```

```python
import matplotlib.pyplot as plt

# Create subplots — 3 rows, 1 column
fig, axs = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

# Plot x1
axs[0].scatter(df['date'], df['meantemp'], color='red', s=10, label='Actual')
axs[0].plot(df['date'], x1, color='black', label='Model x1')
axs[0].set_title('Comparison with x1')
axs[0].legend()

# Plot x2
axs[1].scatter(df['date'], df['meantemp'], color='red', s=10, label='Actual')
axs[1].plot(df['date'], x2, color='black', label='Model x2')
axs[1].set_title('Comparison with x2')
axs[1].legend()

# Plot x3
axs[2].scatter(df['date'], df['meantemp'], color='red', s=10, label='Actual')
axs[2].plot(df['date'], x3, color='black', label='Model x3')
axs[2].set_title('Comparison with x3')
axs[2].legend()

plt.tight_layout()
plt.xlabel('Date')
plt.show()
```
